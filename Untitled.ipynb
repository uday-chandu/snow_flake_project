```python
### Load
"""python
# LOAD: Write optimized Parquet data, partitioned by category
result_df.write.format("parquet") \
    .mode("overwrite") \
    .partitionBy("category") \
    .save("s3a://analytics-bucket/processed_events/")
"""

---

## 2. Streaming Data Handling

PySpark utilizes **Structured Streaming**, a stream processing engine built on the Spark SQL optimization layer. It treats a live data stream as an *unbounded table* that continually appends rows.

### Setting up a Stream (Extract)
When reading streams (e.g., from Kafka or cloud directories), defining an explicit schema is mandatory for most formats.

"""python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType

# Define Schema
schema = StructType([
    StructField("event_id", StringType(), True),
    StructField("user_id", StringType(), True),
    StructField("amount", IntegerType(), True),
    StructField("timestamp", TimestampType(), True)
])

# Stream from a Directory (e.g., Auto Loader / cloudFiles)
streaming_df = spark.readStream \
    .format("json") \
    .schema(schema) \
    .load("s3a://incoming-events-stream/")
"""

### Stateless vs. Stateful Transformations
Understanding the computation types dictates how memory and state are managed during a stream.

| Transformation Type | Description | PySpark Example |
| :--- | :--- | :--- |
| **Stateless** | Operations applied to each row independently. Does not require memory of previous rows. | `filter()`, `select()`, `withColumn()` |
| **Stateful** | Operations that combine data across multiple rows or time intervals. Requires an engine state store. | `groupBy()`, `window()`, `dropDuplicates()` |

#### Managing Late Data with Watermarking
In stateful streaming, data can arrive late due to network latency. **Watermarking** tells Spark how long to wait for late data before closing a time window and cleaning up the state cache.

"""python
from pyspark.sql.functions import window, col

# Apply a 10-minute watermark and aggregate over a 5-minute Tumbling Window
windowed_counts = streaming_df \
    .withWatermark("timestamp", "10 minutes") \
    .groupBy(
        window(col("timestamp"), "5 minutes"),
        col("user_id")
    ).sum("amount")
"""

### Writing the Stream (Load)
To output your stream, you must specify an **Output Mode**, a **Trigger**, and a **Checkpoint Location** to ensure fault-tolerance.

"""python
query = windowed_counts.writeStream \
    .format("delta") \
    .outputMode("complete") \
    .option("checkpointLocation", "s3a://checkpoint-bucket/metrics/") \
    .trigger(processingTime='1 minute') \
    .start("s3a://analytics-bucket/live_metrics/")

query.awaitTermination()
"""

### Key Streaming Configurations
*   **Output Modes:** 
    *   `append`: Default mode; only new rows are added to the sink (required for stateless streams).
    *   `complete`: The entire updated result table is rewritten to the sink (used for global aggregations).
    *   `update`: Only the rows modified since the last micro-batch are written.
*   **Triggers:** Defines *when* data is processed (e.g., `processingTime='10 seconds'`, `once=True` for cost-efficient batch-like intervals, or `availableNow=True`).
*   **Checkpoints:** Saves execution state metadata to cloud storage. If a stream crashes, it resumes exactly where it left off.

---

## 3. Advanced Pattern: `foreachBatch`

Some operations (like upserting into relational databases via JDBC, multi-sink routing, or applying complex deduplication) cannot be done directly inline in structured streaming. The `foreachBatch()` method allows you to execute arbitrary batch DataFrame functions on each micro-batch.

"""python
def merge_into_delta(micro_batch_df, batch_id):
    """
    Treats the micro-batch stream partition as a standard Batch DataFrame.
    Useful for executing Delta Lake MERGE statement (upserts).
    """
    # E.g., remove duplicates within this micro-batch partition before writing
    clean_batch_df = micro_batch_df.dropDuplicates(["event_id"])
    
    # Save or update targeted database
    clean_batch_df.write \
        .format("delta") \
        .mode("append") \
        .save("s3a://analytics-bucket/final_table/")

# Wire the batch function into the streaming write pipeline
query = streaming_df.writeStream \
    .foreachBatch(merge_into_delta) \
    .option("checkpointLocation", "s3a://checkpoint-bucket/merge_logs/") \
    .start()
"""
```

```python
import json
import pandas as pd


def flatten_and_explode_json(json_data):
    # Load JSON if it's a string
    if isinstance(json_data, str):
        data = json.loads(json_data)
    else:
        data = json_data

    # Ensure data is a list of records for processing
    if not isinstance(data, list):
        data = [data]

    # Step 1: Flatten dictionary structures (e.g., customer.address)
    df = pd.json_normalize(data)

    # Step 2: Dynamically find and explode list-like columns (e.g., items, metadata)
    list_cols = [
        col
        for col in df.columns
        if df[col].apply(lambda x: isinstance(x, list)).any()
    ]

    for col in list_cols:
        df = df.explode(col)

    # Step 3: Secondary normalization if the exploded list contained dictionaries
    # (Extracts properties inside 'items' and 'metadata' into their own columns)
    for col in list_cols:
        # Check if the exploded elements are dictionaries
        if df[col].apply(lambda x: isinstance(x, dict)).any():
            nested_df = pd.json_normalize(df[col].dropna().tolist()).set_index(
                df[col].dropna().index
            )
            # Prefix columns to avoid collisions (e.g., items.price)
            nested_df = nested_df.add_prefix(f"{col}.")
            df = df.drop(columns=[col]).join(nested_df)

    return df.reset_index(drop=True)


final_df = flatten_and_explode_json(json_data)
final_df
```

```python
from pyspark.sql.types import StructType, StructField, StringType, LongType, DoubleType, ArrayType

# Define the complete schema matching your printSchema() output
spark_schema = StructType([
    # 1. Customer Nested Struct
    StructField("customer", StructType([
        StructField("address", StructType([
            StructField("city", StringType(), True),
            StructField("country", StringType(), True),
            StructField("postal_code", StringType(), True)
        ]), True),
        StructField("customer_id", LongType(), True),
        StructField("email", StringType(), True),
        StructField("name", StringType(), True)
    ]), True),
    
    # 2. Items Array of Structs
    StructField("items", ArrayType(
        StructType([
            StructField("item_id", StringType(), True),
            StructField("price", DoubleType(), True),
            StructField("product_name", StringType(), True),
            StructField("quantity", LongType(), True)
        ]), True
    ), True),
    
    # 3. Metadata Array of Structs
    StructField("metadata", ArrayType(
        StructType([
            StructField("key", StringType(), True),
            StructField("value", StringType(), True)
        ]), True
    ), True),
    
    # 4. Order ID Root Field
    StructField("order_id", StringType(), True),
    
    # 5. Payment Nested Struct
    StructField("payment", StructType([
        StructField("method", StringType(), True),
        StructField("transaction_id", StringType(), True)
    ]), True)
])
```
