```python
# Test your function with this:
nums = [10, 20, 30, 40, 50]
target = 70
# Expected Output: [2, 3] (because numbers[2] + numbers[3] -> 30 + 40 = 70)

def find_pair(nums, target):
    seen_number={}

    for current_index,current_number in enumerate(nums):
        complement = target-current_number

        if complement in seen_number:
            return seen_number[complement],current_index

        seen_number[current_number]=current_index

    return None
```

```python
def flatten_dict(nested_dict, parent_key='', sep='_'):
    items = {}
    
    for k, v in nested_dict.items():
        # 1. Calculate the new combined key name
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        print(new_key)
        if isinstance(v,dict):
            items.update(flatten_dict(v,new_key,sep=sep))
        else:
            items[new_key]=v
    return items
flatten_dict(nested_dict=nested_data)
```

```python
def compress_status(signals: str) -> str:
    string_output=''
    count=0
    for ind,i in enumerate(signals):
        if signals[ind]==signals[ind-1]:
            count+=1
        else:
            string_output+=signals[ind-1]+str(count)
            count=1
    string_output+=signals[ind-1]+str(count)
    return string_output

# Execution Test
print(compress_status("AAAAABBBCCDAA"))
# Expected Output: "A5B3C2D1A2"
```

```python
raw_logs = [
    {"device_id": "A1", "temperature": 23.5},
    {"device_id": None, "temperature": 15.0},          # Rule 1: Drop (Missing ID)
    {"device_id": "A2", "temperature": None},          # Rule 2: Replace with 20.0
    {"device_id": "A3", "temperature": 150.0},         # Rule 3: Drop (Outlier!)
    {"device_id": "   ", "temperature": 12.0},         # Rule 1: Drop (Empty string ID)
    {"device_id": "A4", "temperature": -12.5}          # Keep
]

def clean_sensor_data(logs):
    # Write your data cleaning logic loop here
    for idx,log in enumerate(logs):
        if log in logs and log['device_id']==None  or len(log['device_id'].strip())==0:
            logs.remove(log)
        if log in logs and log['temperature']==None:
            logs[idx]['temperature']=20.0
        if log in logs and abs(log['temperature'])>=100:
            logs.remove(log)
    return logs
    pass

clean_sensor_data(raw_logs)
# Expected Output:
# [
#   {"device_id": "A1", "temperature": 23.5},
#   {"device_id": "A2", "temperature": 20.0},
#   {"device_id": "A4", "temperature": -12.5}
# ]
```
